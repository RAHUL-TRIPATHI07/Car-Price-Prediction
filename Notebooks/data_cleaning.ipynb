```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\Rahul\Desktop\Codes\aiml\Machine Learning\LinearRegression\CarPricePrediction\Data\cardekho_dataset.csv")
```

```python
df = df.drop(columns=["Unnamed: 0"])

print(df.shape)
```

```python
numeric_cols = [
    "vehicle_age",
    "km_driven",
    "mileage",
    "engine",
    "max_power",
    "seats",
    "selling_price"
]

df[numeric_cols].describe().T
```

```python
for col in numeric_cols:
    print(f"{col}: {(df[col] <= 0).sum()} non-positive values")
```

```python
categorical_cols = [
    "brand",
    "model",
    "seller_type",
    "fuel_type",
    "transmission_type"
]

for col in categorical_cols:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

```python
df[["car_name", "brand", "model"]].head(10)
```

```python
print("Unique car names:", df["car_name"].nunique())
print("Unique brand-model combinations:",
      df.groupby(["brand", "model"]).ngroups)
```

```python
# Remove redundant car_name column
df = df.drop(columns=["car_name"])

print("Shape after cleaning:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicates:", df.duplicated().sum())
```

```python
df = df.drop_duplicates().reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)
print("Duplicates remaining:", df.duplicated().sum())
```
