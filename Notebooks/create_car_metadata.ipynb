```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\Rahul\Desktop\Codes\aiml\Machine Learning\LinearRegression\CarPricePrediction\Data\cardekho_dataset.csv")
```

```python
car_mapping = (
    df[["brand", "model"]]
    .drop_duplicates()
    .sort_values(["brand", "model"])
)

print(car_mapping)
```

```python
import json

brand_model_mapping = (
    car_mapping
    .groupby("brand")["model"]
    .apply(list)
    .to_dict()
)

with open("brand_model_mapping.json", "w") as f:
    json.dump(brand_model_mapping, f, indent=4)
```
